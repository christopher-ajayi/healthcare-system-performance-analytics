### **3.1 Nominal Expenditure Analysis (1975-2025)**

```python
df_expenditure = (
    df_expenditure
       .sort_values(["province_code", "data_year"])
    .reset_index(drop=True)
)
```

```python
df_expenditure.head(10)
```

```python
df_expenditure["expenditure_growth_pct"] = (
    df_expenditure
    .sort_values(["province_code", "data_year"])
    .groupby("province_code")["total_health_expenditure_current"]
    .pct_change(fill_method=None) * 100
)
```

```python
df_norm_expenditure_change = (
    df_expenditure[
        df_expenditure["total_health_expenditure_current"].notna()
    ]
    .sort_values(["province_code", "data_year"])
    .groupby(["province_code", "province_name"])
    .agg(
        start_year=("data_year", "first"),
        end_year=("data_year", "last"),
        start_expenditure=("total_health_expenditure_current", "first"),
        end_expenditure=("total_health_expenditure_current", "last")
    )
    .reset_index()
)
```

```python
df_norm_expenditure_change["expenditure_growth_pct"] =(
    (df_norm_expenditure_change["end_expenditure"]
    - df_norm_expenditure_change["start_expenditure"])
    / df_norm_expenditure_change["start_expenditure"]
) * 100
```

```python
df_norm_expenditure_change
```

```python
df_norm_expenditure_change.sort_values("expenditure_growth_pct", ascending=False)
```

### **3.2 Real Expenditure Growth (1975-2025)**

```python
df_expenditure["real_expenditure_growth_pct"] = (
    df_expenditure
    .sort_values(["province_code", "data_year"])
    .groupby("province_code")["total_health_expenditure_constant_2010"]
    .pct_change(fill_method=None) * 100
)
```

```python
df_real_expenditure_change = (
    df_expenditure[
        df_expenditure["total_health_expenditure_constant_2010"].notna()
    ]
    .sort_values(["province_code", "data_year"])
    .groupby(["province_code", "province_name"])
    .agg(
        start_year=("data_year", "first"),
        end_year=("data_year", "last"),
        start_real_expenditure=("total_health_expenditure_constant_2010", "first"),
        end_real_expenditure=("total_health_expenditure_constant_2010", "last")
    )
    .reset_index()
)
```

```python
df_real_expenditure_change["real_expenditure_growth_pct"]= (
    (df_real_expenditure_change["end_real_expenditure"]
    - df_real_expenditure_change["start_real_expenditure"])
    / df_real_expenditure_change["start_real_expenditure"]
) * 100
```

```python
df_real_expenditure_change.sort_values("real_expenditure_growth_pct", ascending=False)
```

```python
df_public_private = df_expenditure[
    [
        "province_code",
        "province_name",
        "data_year",
        "public_share_pct",
        "private_share_pct"
    ]
].dropna(
    subset=["public_share_pct", "private_share_pct"]
).sort_values(["province_code", "data_year"]

).reset_index(drop=True)
```

> - ##### Overall, the results indicate a shift toward a relatively larger private share of healthcare expenditure in most province over the period. Six of the ten provinces experienced a decline in the public share, while four experienced an increase. This suggests that the financing composition of healthcare changed over time, although the extent of the shift varied considerably across jurisdictions.
> - ##### From a public-finance perspective, these differences are relevant because the composition of healthcare financing indicates how financial responsibility is distributed between public and private sources. A declining public share does not by itself indicate weaker public healthcare provision or poorer system performance, but it provides an important basis for examining government expenditure, resource allocation, and financial sustainability in the subsequent analysis.

### **3.6 How much has healthcare expenditure changed relative to population and demographic ageing?**

```python
#### Recalled from Notebook 2 (chps_resources) (RQ2 - Population)

query = """
    SELECT
    province_id,
    province_code,
    province_name,
    data_year,
    population,
    population_65_plus,
    population_80_plus,
    population_85_plus,
    population_65_share_pct,
    population_80_share_pct,
    population_85_share_pct
FROM master.analytics_province_year
WHERE province_code NOT IN ('CA', 'NT', 'YT', 'NU')
ORDER BY province_name, data_year;
"""

df_population_demand = pd.read_sql(query, engine)

df_population_demand
df_population_demand["population_growth_pct"] = (
    df_population_demand
    .groupby('province_code')["population"]
    .pct_change() * 100
)

df_population_demand
df_population_demand[
    [
        "province_name",
        "province_code",
        "population",
        "population_growth_pct"
    ]
    
].head()
df_population_demand["population_growth_pct"].describe()
df_population_change = (
    df_population_demand
    .sort_values(["province_code", "data_year"])
    .groupby(["province_code", "province_name"])
    .agg(
        start_year=("data_year", "min"),
        end_year=("data_year", "max"),
        start_population=("population", "first"),
        end_population=("population", "last")
        
    )
    .reset_index()
)
df_population_change
df_population_change["absolute_change"] =(
    df_population_change["end_population"] - 
    df_population_change["start_population"]
)

df_population_change["population_change_pct"] = (
    df_population_change["absolute_change"]
    / df_population_change["start_population"]
) * 100

df_population_change
```

```python
df_population_demand_1975 = df_population_demand[
    (df_population_demand["data_year"].isin([1975, 2025]) & 
    ~(df_population_demand["province_code"]).isin(["CA", "NU", "NT", "YT"]))
].copy()
```

```python
df_population_demand_1975 = (
    df_population_demand_1975
    .pivot(
        index=["province_code", "province_name"],
        columns="data_year",
        values=["population", "population_65_plus"]
    )
    .reset_index()
)
```

```python
df_population_demand_1975["population_65_growth_pct"] = (
    (df_population_demand_1975["population_65_2025"]
    - df_population_demand_1975["population_65_1975"])
    / df_population_demand_1975["population_65_1975"]
) * 100
```

```python
df_expenditure_demand["expenditure_population_gap_pp"] = (
    df_expenditure_demand["real_expenditure_growth_pct"]
    - df_expenditure_demand["population_growth_pct"])
```

```python
df_expenditure_demand["expenditure_ageing_gap_pp"] = (
    df_expenditure_demand["real_expenditure_growth_pct"]
    - df_expenditure_demand["population_65_growth_pct"]
)
```

```python
df_expenditure_demand[
    [
        "province_code",
        "province_name",
        "real_expenditure_growth_pct",
        "population_growth_pct",
        "population_65_growth_pct",
        "expenditure_population_gap_pp",
        "expenditure_ageing_gap_pp"
    ]
].sort_values("expenditure_population_gap_pp", ascending=False)
```

> ##### **Insights:** 
> - ##### Real healthcare expenditure increased substantially faster than population growth in all ten provinces between 1975 and 2025. The largest expenditure–population gap was recorded in Alberta, where real healthcare expenditure increased by 618.0%, compared with population growth of 178.1%, producing a gap of +439.9 percentage points. British Columbia recorded a similar pattern, with real expenditure increasing by 563.3% compared with population growth of 127.9%, resulting in a +435.4 percentage-point gap. Even the smallest gap, in Quebec, was +246.9 percentage points, reflecting real expenditure growth of 290.0% compared with population growth of 43.1%.
> - ##### The comparison with population ageing presents a more differentiated picture. In nine of the ten provinces, real healthcare expenditure grew faster than the population aged 65+. Manitoba recorded the largest expenditure–ageing gap at +185.4 percentage points, with real expenditure increasing by 335.3% compared with 149.8% growth in its 65+ population. British Columbia recorded a +170.7 percentage-point gap, while Saskatchewan recorded +172.6 percentage points. Quebec was the only province where ageing growth exceeded real expenditure growth: its 65+ population increased by 316.9%, compared with 290.0% real expenditure growth, producing a −26.9 percentage-point gap.
> - ##### From a healthcare-sustainability perspective, these results indicate that real healthcare expenditure expanded considerably relative to overall population growth across all provinces and generally also exceeded the growth of the older population. However, the magnitude of these expenditure–demand gaps should not be interpreted as evidence that resources were sufficient or that healthcare performance improved. They indicate the relative pace of financial resource expansion compared with demographic change and provide a basis for examining whether expenditure growth corresponds with changes in healthcare capacity, access, and outcomes.


### **3.7 How does expenditure relate to healthcare capacity?**

```python
query = """
SELECT
    Province_id,
    province_code,
    province_name,
    data_year,
    population,
    population_65_plus,
    population_80_plus,
    population_85_plus,
    population_65_share_pct,
    population_80_share_pct,
    population_85_share_pct,
    hospital_beds_total,
    current_hospital_beds_total,
    physicians_total,
    registered_nurses,
    licensed_practical_nurses,
    nurse_practitioners,
    registered_psychiatric_nurses,
    icu_beds,
    long_term_care_beds,
    mental_health_addictions_beds,
    obstetrics_beds,
    other_acute_care_beds,
    pediatrics_beds,
    rated_capacity_beds,
    rehabilitation_beds
FROM master.analytics_province_year
WHERE province_code NOT IN ('CA', 'NT', 'YT', 'NU')
ORDER BY province_name, data_year;
"""

df_healthcare_capacity = pd.read_sql(query, engine)

#### Changes in Bed Capacity Across Provinces
df_bed_capacity_change = (df_healthcare_capacity[
  df_healthcare_capacity["hospital_beds_total"].notna()
].groupby(
    ["province_code", "province_name"]
).agg(
    start_year=("data_year", "min"),
    end_year=("data_year", "max"),
    start_beds=("hospital_beds_total", "first"),
    end_beds=("hospital_beds_total", "last")
).reset_index()
)

##### Absolute Change in Bed Capacity
df_bed_capacity_change["bed_change_absolute"] = (
    df_bed_capacity_change["end_beds"] - df_bed_capacity_change["start_beds"]
)


 #### Changes in Bed Capacity Across Provinces
df_bed_capacity_change["bed_change_pct"] = (
    df_bed_capacity_change["bed_change_absolute"]
    / df_bed_capacity_change["start_beds"]
) * 100
```

```python
df_expenditure_capacity = (
    df_expenditure[
    (df_expenditure["data_year"].isin([2009, 2022]))
    ]
    .sort_values(["province_code", "data_year"])
    .groupby(["province_code", "province_name"])
    .agg(
    start_real_expenditure=("total_health_expenditure_constant_2010", "first"),
    end_real_expenditure=("total_health_expenditure_constant_2010", "last")
).reset_index()

)

df_expenditure_capacity
```

```python
df_expenditure_capacity["real_expenditure_growth_pct"] = (
    (df_expenditure_capacity["end_real_expenditure"] - df_expenditure_capacity["start_real_expenditure"]) / df_expenditure_capacity["start_real_expenditure"]
)* 100

df_expenditure_capacity
```

```python
df_expenditure_resource = df_expenditure_capacity.merge(
    df_bed_capacity_change[
        [
            "province_code",
            "start_year",
            "end_year",
            "bed_change_pct"
        ]
    ],
    on="province_code",
    how="inner"
)
```

```python
df_expenditure_resource
```

```python
df_expenditure_resource["expenditure_capacity_gap"] = (

    df_expenditure_resource["real_expenditure_growth_pct"] 
    - df_expenditure_resource["bed_change_pct"]
)
```

```python
df_expenditure_resource.sort_values("expenditure_capacity_gap", ascending=False)
```

British Columbia recorded the larget expenditure-capacity gap, with real healthcare expenditure increasing by 60.8% while hospital beds declined by 18.3%, producing a 77.1 pecentage-point gap. Prince Edward Island recorded the second largest gap at 61.4 percentage points, while Ontario recorded the smallest at 26.6 percentage points, with expenditure increasing by 36.9% compared with 10.3% growth in beds. This shows that growth in financial resources exceeded growth in hospital capacity. This is probably because additional healthcare expenditure could mean spending on other health related priorities like workforce and access, which are beyond hospital-bed capacity expansion.
